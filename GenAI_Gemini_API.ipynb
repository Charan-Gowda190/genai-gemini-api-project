````python
from google import genai
from google.colab import userdata

# Store GOOGLE_API_KEY in Google Colab Secrets. Never hard-code API keys.
api_key = userdata.get("GOOGLE_API_KEY").strip()
client = genai.Client(api_key=api_key)

print("Gemini client created successfully!")
````

````python
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Explain Artificial Intelligence in simple words"
)

print(response.text)
````

## Model Parameters

Model parameters influence how the AI generates a response.

### Temperature
- **Low (0–0.3):** more predictable and consistent output; useful for coding and precise tasks.
- **Medium (0.4–0.7):** balanced output; useful for factual/general tasks.
- **High (0.8–1.0):** more creative and varied output; useful for writing and content generation.

### Max Output Tokens
Controls the maximum length of the generated response.


````python
response = client.models.generate_content(
    model = "gemini-2.5-flash",
    contents = "Write a short story",
    config = {'temperature':0.9, "max_output_tokens":100}  #### generates new content
)
print(response.text)
````

````python
from google.genai import types

text = """
Summarize the following text -
The two new Delhi Metro corridors inaugurated by the Prime Minister were the Majlis Park-Maujpur Babarpur (Pink Line) corridor of around 12.3 km and the Deepali Chowk-Majlis Park (Magenta Line) corridor of around 9.9 km.
These corridors will benefit several areas of Delhi, including Burari, Jagatpur-Wazirabad, Khajuri Khas, Bhajanpura, Yamuna Vihar, Madhuban Chowk, Haiderpur Badli Mor, Bhalaswa and Majlis Park.
"""

response = client.models.generate_content(
    model = "gemini-2.5-flash",
    contents = [
        types.Content(
            role = "user",
            parts = [
                types.Part.from_text(text = text)
            ]
        )
            ]
        )
````

````python
print(response.text)
````

````python
# System prompt
system_prompt = "You are a helpful assistant that summarizes text in a short and clear way."

# User prompt
user_prompt = """
Summarize the following text:

The two new Delhi Metro corridors inaugurated by the Prime Minister were the Majlis Park-Maujpur Babarpur (Pink Line) corridor of around 12.3 km and the Deepali Chowk-Majlis Park (Magenta Line) corridor of around 9.9 km.
These corridors will benefit several areas of Delhi, including Burari, Jagatpur-Wazirabad, Khajuri Khas, Bhajanpura, Yamuna Vihar, Madhuban Chowk, Haiderpur Badli Mor, Bhalaswa and Majlis Park.
"""

# Generate response
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=f"{system_prompt}\n\n{user_prompt}"
)

print(response.text)
````
